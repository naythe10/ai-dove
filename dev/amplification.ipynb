```python
# Boost after capture while preventing int16 overflow.
AMPLIFICATION_DB = 12
amplification = 10 ** (AMPLIFICATION_DB / 20)

raw_samples = np.frombuffer(five_sec_audio.data, dtype=np.int16)
boosted_samples = np.clip(
    raw_samples.astype(np.float32) * amplification,
    np.iinfo(np.int16).min,
    np.iinfo(np.int16).max,
).astype(np.int16)

print(
    f"boost={AMPLIFICATION_DB} dB | "
    f"raw_peak={np.max(np.abs(raw_samples))} | "
    f"boosted_peak={np.max(np.abs(boosted_samples))}"
)
```
